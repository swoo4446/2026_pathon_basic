## 1. 표준 출력 `print()`

- `print()`는 값을 화면에 출력할 때 사용
- 문자열은 따옴표로 감싸고, 숫자는 그대로 사용
- 변수에 저장된 값도 출력 가능


````python
# 문자열과 변수 출력
print("이성우")

name = "이성우"
print(name)

# 문자열과 숫자는 서로 다른 자료형
text_number = "123"
number = 123

print(text_number)
print(number)
````

### 문자열 포매팅

- `format()` 또는 f-string으로 변수 값을 문장 안에 삽입
- 실무에서는 가독성이 좋은 **f-string**을 자주 사용


````python
name = "이성우"
age = 35

# format() 메서드
print("이름: {}, 나이: {}".format(name, age))

# f-string
print(f"이름: {name}, 나이: {age}")
print(f"안녕하세요, 제 이름은 {name}이고 나이는 {age}살입니다.")
````

### `sep`, `end`, `flush`

- `sep`: 여러 출력값 사이의 구분자 설정
- `end`: 출력 마지막 문자를 설정
- `flush=True`: 출력 버퍼를 즉시 비워 진행 상태처럼 바로 갱신할 때 활용


````python
# sep: 출력값 사이의 구분자
print("A", "B", "C")
print("A", "B", "C", sep="-")
print("2025", "12", "21", sep="/")

# end: 줄바꿈 대신 다른 문자 사용
print("로딩 중", end="...")
print("완료!")

# 반복 출력
for _ in range(3):
    print(".", end="")
print()

# 여러 값을 다른 줄에 출력
print(100, 200, 300, 400, 600, sep="\n   &   ", end=" end\n")
````

````python
import time

# 같은 줄의 값을 갱신하는 예시
for i in range(5):
    print(f"진행: {i + 1}", end="\r", flush=True)
    time.sleep(1)

print("진행 완료!")
````

## 5. 리스트

- 여러 값을 순서대로 저장하는 자료형
- 인덱스는 `0`부터 시작
- 요소 변경, 중첩 리스트, 리스트 연산 가능


````python
fruits = ["사과", "바나나", "포도"]
numbers = [1, 2, 3, 4, 5]
mixed = [10, "Python", 3.14, True]

# 요소 변경
numbers[1] = 100
fruits[2] = "오렌지"

print(numbers)
print(fruits)
print(type(fruits[0]))

# 문자열도 인덱싱 가능
print(fruits[1][0])
````

## 6. 리스트·셋·딕셔너리

- **리스트(list)**: 순서가 있고 수정 가능
- **셋(set)**: 중복을 허용하지 않는 집합
- **딕셔너리(dict)**: `key: value` 형태로 데이터 저장


````python
# 리스트
fruits = ["사과", "바나나", "사과"]
print("list:", fruits)

# 셋: 중복 제거
unique_fruits = set(fruits)
print("set:", unique_fruits)

# 딕셔너리: key-value 구조
student = {
    "name": "이성우",
    "age": 35,
    "course": "Python"
}

print("dict:", student)
print("이름:", student["name"])
````

## 7. 연습문제: 간식 상자

### 초기 시도에서 확인한 포인트
- `append()`는 값을 추가하지만 반환값은 `None`
- `input()` 결과는 문자열이므로 숫자로 사용할 때 `int()` 변환 필요
- `remove()`는 **값**을 삭제하고, `pop()`은 **인덱스**로 삭제
- 여러 간식을 한 번에 추가할 때는 `append()`보다 `extend()`가 자연스러움


### 잘못되기 쉬운 예시

```python
snack_box.remove(input('빼고 싶은 간식의 번호를 입력하세요 : ') - 1)
```

- `input()`은 문자열이므로 바로 `- 1` 연산 불가
- 입력값이 번호라면 `int()` 변환 후 `pop(index)` 사용이 적절


````python
# 개선한 간식 상자 프로그램
snack_box = ["초코파이"]

# 간식 추가
new_snacks = [
    snack.strip()
    for snack in input("먹고 싶은 간식을 쉼표로 구분해 입력하세요: ").split(",")
]

box_count = int(input("몇 상자를 주문하시겠어요?: "))

for _ in range(box_count):
    snack_box.extend(new_snacks)

print("주문 목록:", snack_box)

# 간식 제거
number = int(input("빼고 싶은 간식의 번호를 입력하세요: "))

if 1 <= number <= len(snack_box):
    removed_snack = snack_box.pop(number - 1)
    print(f"{removed_snack}을(를) 제외했습니다.")
else:
    print("올바른 번호를 입력해주세요.")

print("최종 주문 목록:", snack_box)
print("감사합니다.")
````

## 8. 조건문

- `if`로 조건을 검사
- 여러 조건은 `elif`, 나머지는 `else`
- `in`을 사용하면 특정 값이 목록에 포함되는지 확인 가능


````python
score = 50

if score >= 90:
    print("합격입니다.")
else:
    print("불합격입니다.")
````

````python
age = int(input("나이: "))

if age > 20:
    print("환영합니다.")
elif age > 10:
    print("반쯤 환영합니다.")
else:
    print("다음에 만나요.")
````

````python
age = int(input("나이: "))

if age <= 14 or age >= 65:
    print("무료입니다.")
elif age <= 35:
    print("요금은 1,000원입니다.")
else:
    print("요금은 2,000원입니다.")
````

````python
last_number = 5

if last_number in [0, 2, 4, 6, 8]:
    print("짝수")
else:
    print("홀수")
````

## 9. 반복문

- `for`: 정해진 범위나 데이터 요소를 순회
- `range()`: 반복 횟수를 지정할 때 자주 사용
- `while`: 조건이 참인 동안 반복


````python
fruits = ["사과", "바나나", "포도"]

for fruit in fruits:
    print(fruit)
````

````python
# 문자열의 각 문자 순회
text = "hell o"

for char in text:
    print(char)
````

````python
# range(3) -> 0, 1, 2
for x in range(3):
    print(x)
````

````python
# 별 트리 만들기
height = int(input("트리 높이를 입력하세요: "))

for i in range(1, height + 1):
    spaces = " " * (height - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
````

````python
# exit를 입력할 때까지 반복
user_input = ""

while user_input != "exit":
    user_input = input("원하는 과자를 입력하세요 (exit 입력 시 종료): ")

    if user_input != "exit":
        print("입력한 값:", user_input)

print("프로그램을 종료합니다.")
````

## 10. 함수

- 반복해서 사용할 로직을 하나의 이름으로 정의
- `return`으로 함수 실행 결과를 반환


````python
def add_int(a, b):
    # 두 정수를 더한 결과 반환
    result = a + b
    return result

print(add_int(2, 4))
````
