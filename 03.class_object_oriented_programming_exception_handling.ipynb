## 1. 클래스와 객체지향 개념

- **변수**: 데이터를 저장
- **함수**: 특정 동작을 묶어 재사용
- **모듈**: 여러 함수나 클래스를 하나의 `.py` 파일로 구성
- **클래스**: 데이터(변수)와 기능(함수)을 하나의 구조로 묶음
- **객체(인스턴스)**: 클래스를 기반으로 실제 생성한 값

즉, 클래스는 **객체를 만들기 위한 설계도**라고 이해할 수 있습니다.


## 2. 클래스 정의와 객체 생성

- `class` 키워드로 클래스 정의
- `__init__()`은 객체 생성 시 자동 실행되는 초기화 메서드
- `self`는 생성된 객체 자기 자신을 의미
- `self.name`, `self.hobby`처럼 객체별 데이터를 저장


````python
#호출
class Student:
    def __init__(self, name, hobby):
        # 객체별 데이터 저장
        self.name = name
        self.hobby = hobby

    def introduce(self):
        # 객체에 저장된 값을 이용해 자기소개 출력
        print(
            f"안녕, 날 소개하지. "
            f"내 이름은 {self.name}, 취미는 {self.hobby}"
        )
````

````python
# Student 클래스로 객체 생성
s1 = Student("김하온", "독서, 영화시청")

# 객체의 메서드 호출
s1.introduce()
````

## 3. 메서드 활용

- 클래스 내부에 정의한 함수를 메서드(method)라고 함
- 메서드는 객체가 가진 데이터를 이용해 동작 수행
- 같은 클래스로 만든 여러 객체가 동일한 기능을 재사용 가능


## 4. 실습: 계산기 클래스

- `number_list`: 객체가 보유하는 데이터
- `sum()`: 합계 반환
- `avg()`: 평균 반환


````python
class Calculator:
    def __init__(self, number_list):
        self.number_list = number_list

    def sum(self):
        return sum(self.number_list)

    def avg(self):
        return sum(self.number_list) / len(self.number_list)


cal1 = Calculator([1, 2, 3, 4, 5])

print("cal1 합계:", cal1.sum())   # 15
print("cal1 평균:", cal1.avg())   # 3.0

cal2 = Calculator([6, 7, 8, 9, 10])

print("cal2 합계:", cal2.sum())   # 40
print("cal2 평균:", cal2.avg())   # 8.0
````

## 5. 실습: 계좌 클래스

초기 버전의 핵심:
- 은행명, 예금주, 잔액을 객체에 저장
- 계좌번호를 난수로 생성
- 객체마다 서로 다른 계좌 정보를 보유


````python
import random


class AccountBasic:
    def __init__(self, name, balance):
        self.bank = "SC은행"
        self.name = name
        self.balance = balance

        # 계좌번호 랜덤 생성
        num1 = random.randint(0, 999)
        num2 = random.randint(0, 99)
        num3 = random.randint(0, 999999)

        self.account_number = f"{num1:03d}-{num2:02d}-{num3:06d}"


account1 = AccountBasic("홍길동", 10000)

print(f"이름: {account1.name}")
print(
    f"은행: {account1.bank}, "
    f"계좌번호: {account1.account_number}, "
    f"잔액: {account1.balance:,}원"
)
````

## 6. 클래스 변수와 정적 메서드

- **클래스 변수**: 모든 객체가 공통으로 사용하는 값
- **정적 메서드**: 객체 상태와 직접 관계없는 기능을 클래스 내부에 정의
- **인스턴스 메서드**: `self`를 통해 객체 상태를 읽거나 변경


````python
import random


class Account:
    # 모든 Account 객체가 공유하는 클래스 변수
    bank_name = "SC은행"

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        self.account_no = Account.make_account_no()

    @staticmethod
    def make_account_no():
        # 객체 상태와 무관하게 계좌번호 생성
        num1 = random.randint(0, 999)
        num2 = random.randint(0, 99)
        num3 = random.randint(0, 999999)

        return f"{num1:03d}-{num2:02d}-{num3:06d}"

    def print_info(self):
        return (
            f"은행이름: {Account.bank_name}, "
            f"예금주: {self.name}, "
            f"계좌번호: {self.account_no}, "
            f"잔액: {self.balance:,}원"
        )

    def 입금(self, amount):
        self.balance += amount
        return self.balance

    def 출금(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("잔액이 부족합니다.")

        return self.balance

    def 잔고조회(self):
        return self.balance
````

````python
# 첫 번째 계좌 객체
ac1 = Account("홍길동", 2000)

print(ac1.print_info())
print("새 계좌번호 생성 예시:", ac1.make_account_no())
````

````python
# 두 번째 계좌 객체
ac2 = Account("희동이", 5000)

print(ac2.print_info())
print("현재 잔액:", ac2.잔고조회())
````

## 8. 상속과 메서드 재정의

- **상속(inheritance)**: 기존 클래스의 구조와 기능을 새로운 클래스가 이어받는 것
- 부모 클래스의 메서드를 자식 클래스에서 다시 정의할 수 있음
- 이를 **메서드 오버라이딩(재정의)**이라고 함


````python
class Animal:
    def make_sound(self):
        # 자식 클래스에서 구체적으로 구현하도록 비워 둠
        pass


class Dog(Animal):
    def make_sound(self):
        return "멍멍!"


class Cat(Animal):
    def make_sound(self):
        return "야옹!"
````

## 9. 다형성

- 서로 다른 객체가 같은 이름의 메서드를 호출
- 객체의 실제 클래스에 따라 서로 다른 결과 반환
- 같은 반복문 안에서도 객체별 동작을 다르게 처리 가능


````python
animals = [Dog(), Cat()]

for animal in animals:
    print(animal.make_sound())
    print("객체 타입:", type(animal))
````

## 10. 예외처리

### 문법 오류 예시

원본 수업 코드:

```python
while True
    print("hello")
```

- `while True` 뒤에 `:`가 없어 `SyntaxError` 발생
- 문법 오류는 프로그램 실행 전에 수정해야 함

수정하면 다음과 같습니다.

```python
while True:
    print("hello")
```

> 위 코드는 무한 반복되므로 실제 실행하지 않는 것이 안전


### `try-except`로 실행 중 오류 처리

- `ValueError`: 숫자로 변환할 수 없는 값을 입력했을 때
- `ZeroDivisionError`: `0`으로 나누었을 때
- 오류가 발생해도 프로그램이 바로 종료되지 않도록 예외 처리


````python
try:
    num = int(input("숫자를 입력하세요: "))
    result = 10 / num
    print("계산 결과:", result)

except ZeroDivisionError:
    # 0으로는 나눌 수 없음
    print("0을 제외한 숫자를 입력하세요.")

except ValueError:
    # int()로 변환할 수 없는 값 입력
    print("숫자만 입력해주세요.")
````

### 사용자 정의 예외

- 사용자정의 예외 클래스 정의 > 예외 발생 > 예외 처리


````python
#예외 처리
try :
    raise TypeError('입력값 오류')

except :
    print('괜찮아요')


try :
    raise TypeError('입력값 오류')

except TypeError as te :
    print(str(te) + ' 괜찮아요')
    print(te.args, ' 괜찮아요')
    print(te.args[0] + ' 괜찮아요')

#예외 클래스 정의

class AgeException(Exception):
    def __init__(self, msg):
        self.message = msg

def input_age():
    age = int(input("나이를 입력하세요: "))

    if age < 0:
        raise AgeException("나이는 0 이상이어야 합니다.")

    elif age >= 150:
        raise AgeException("나이는 150이 넘을 수 없습니다.")

    else:
        return age

# 예외처리를 하는 나이 입력 구분

class AgeException(Exception):
    def __init__(self, msg):
        self.message = msg

def input_age():
    try :
        age = int(input("나이를 입력하세요: "))

        if age < 0:
            raise AgeException("나이는 0 이상이어야 합니다.")

        elif age >= 150:
            raise AgeException("나이는 150이 넘을 수 없습니다.")

        else:
            return age
    
    except AgeException as e:
        print(f"나이 입력 오류: {e}")
    
    except ValueError:
        print("나이는 숫자로 입력해주세요.")

input_age()
````
