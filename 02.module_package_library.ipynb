## 1. 모듈 · 패키지 · 라이브러리 개념

### 모듈(Module)
- 함수, 변수, 클래스 등을 담고 있는 **하나의 `.py` 파일**
- 코드를 기능별로 나누어 재사용할 때 활용
- 예: `calc.py`, Python의 `math` 모듈

### 패키지(Package)
- 서로 관련된 **여러 모듈을 하나의 폴더로 묶은 구조**
- 기능이 많아질 때 코드를 역할별로 분리해 관리하기 좋음
- 일반적으로 패키지 폴더 안에 `__init__.py`를 두어 패키지임을 명확하게 표시
- 예: `my_package/calculator.py`, `my_package/message.py`

### 라이브러리(Library)
- 특정 목적을 위해 만들어진 **모듈과 패키지의 모음**
- 다른 개발자가 만든 기능을 설치하거나 불러와 재사용
- 예: `pandas`, `requests`

### 한 번에 비교하기

| 구분 | 의미 | 예시 |
|---|---|---|
| 모듈 | 하나의 Python 파일 | `calc.py` |
| 패키지 | 여러 모듈을 묶은 폴더 | `my_package/` |
| 라이브러리 | 특정 목적의 모듈·패키지 모음 | `pandas`, `requests` |

> 핵심 흐름: **함수/클래스 → 모듈(.py) → 패키지(폴더) → 라이브러리(기능 모음)**


## 2. 사용자 정의 모듈 `calc.py`

직접 만든 Python 파일도 하나의 **모듈**로 사용할 수 있습니다.

`calc.py` 파일:

```python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b
```

### 파일 위치

Notebook과 `calc.py`를 같은 폴더에 둡니다.

```text
python-study/
├─ 02_module_package_library_updated.ipynb
├─ calc.py
└─ my_package/
```

그러면 Notebook에서 다음처럼 불러올 수 있습니다.

```python
import calc
print(calc.add(3, 5))
```


## 3. 패키지 만들기 `my_package`

모듈이 하나의 `.py` 파일이라면, **패키지는 여러 모듈을 하나의 폴더로 묶은 구조**입니다.

### 권장 폴더 구조

```text
python-study/
├─ 02_module_package_library_updated.ipynb
├─ calc.py
│
└─ my_package/
   ├─ __init__.py
   ├─ calculator.py
   └─ message.py
```

### 각 파일의 역할
- `__init__.py`: 이 폴더가 Python 패키지라는 것을 명확히 표시
- `calculator.py`: 덧셈·곱셈 기능
- `message.py`: 인사말 생성 기능

> `my_package` 폴더는 Notebook 파일과 **같은 폴더 안**에 넣으면 아래 import 예제를 바로 실행 가능


### `my_package/calculator.py`

```python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b
```

### `my_package/message.py`

```python
def hello(name):
    return f"안녕하세요, {name}님!"
```

### `my_package/__init__.py`

```python
from .calculator import add, multiply
from .message import hello
```

`__init__.py`에서 필요한 기능을 미리 가져오면 사용자는 패키지 내부 구조를 모두 알지 않아도 사용 가능

````python
# 패키지의 특정 모듈에서 함수 불러오기
from my_package.calculator import add, multiply

print("3 + 5 =", add(3, 5))
print("4 × 6 =", multiply(4, 6))
````

````python
# __init__.py에서 노출한 함수는 패키지에서 바로 불러올 수 있음
from my_package import hello

print(hello("이성우"))
````

### 패키지를 사용하는 이유

- 파일 하나가 너무 커지는 것을 방지
- 기능별로 코드를 분리하여 관리
- 필요한 기능만 선택해서 import 가능
- 프로젝트 규모가 커져도 구조를 이해하기 쉬움

예를 들어 웹 프로젝트라면 다음처럼 역할별 패키지로 나눌 수 있습니다.

```text
project/
├─ models/
├─ services/
├─ utils/
└─ tests/
```

즉, 패키지는 단순히 폴더를 만드는 것이 아니라 **관련 기능을 구조적으로 관리하는 방법**


## 4. `import` 사용 방법

### 4-1. 모듈 전체 불러오기

- `import 모듈명` 형태로 사용
- 함수 호출 시 `모듈명.함수명()`으로 출처가 명확하게 표시됨


````python
import calc

# calc 모듈의 add 함수 사용
result = calc.add(3, 5)
print(result)
````

### 4-2. 별칭(alias) 사용하기

- `as`를 사용해 모듈 이름을 짧게 지정
- 이름이 긴 라이브러리를 반복해서 사용할 때 편리


````python
import calc as cc

print(cc.add(10, 20))
````

### 4-3. 필요한 함수만 가져오기

- `from 모듈 import 함수` 형태
- 필요한 함수만 직접 호출 가능


````python
from calc import add, multiply

print("10 × 5 =", multiply(10, 5))
print("2 + 3 =", add(2, 3))
````

### 4-4. `import *` 이해하기


```python
from calc import *
```

- 모듈의 여러 이름을 현재 공간으로 가져오는 방식
- 어떤 함수가 어디에서 왔는지 파악하기 어려워질 수 있음
- 학습 목적으로 이해하되, 포트폴리오 코드에서는 필요한 이름을 명시하는 편이 읽기 쉬움


## 5. Python 표준 라이브러리


| 모듈 | 주요 용도 |
|---|---|
| `math` | 수학 계산 |
| `random` | 난수 생성 |
| `datetime` | 날짜·시간 처리 |
| `os` | 운영체제 및 경로 관련 기능 |
| `sys` | Python 실행 환경 정보 |
| `time` | 시간 관련 기능 |


````python
# math: 수학 함수
import math

print("16의 제곱근:", math.sqrt(16))
````

````python
# random: 랜덤값 생성
import random

print("1~10 사이의 난수:", random.randint(1, 10))
````

````python
# datetime: 날짜와 시간
import datetime

now = datetime.datetime.now()
print("현재 날짜와 시간:", now)
````

````python
# os: 현재 작업 폴더 확인
import os

print("현재 작업 폴더:", os.getcwd())
````

````python
# sys: Python 실행 환경 확인
import sys

print("Python 버전:")
print(sys.version)
````

````python
# time: 일정 시간 대기
import time

time.sleep(1)
print("1초 후 실행")
````

## 6. 외부 라이브러리 `requests`

- 웹 서버에 HTTP 요청을 보낼 때 사용하는 외부 라이브러리
- `requests.get()`으로 지정한 URL에 GET 요청
- 응답 객체의 `status_code`로 요청 성공 여부 확인 가능

> 외부 라이브러리는 환경에 따라 별도 설치가 필요하며 인터넷 연결이 필요


````python
import requests

url = "https://www.naver.com/"

try:
    response = requests.get(url, timeout=10)
    print("응답 상태 코드:", response.status_code)
except requests.RequestException as error:
    print("요청 중 오류 발생:", error)
````

## 7. 외부 라이브러리 `pandas`

- 표 형태의 데이터를 다룰 때 사용하는 데이터 분석 라이브러리
- 관례적으로 `pandas`를 `pd`라는 별칭으로 사용
- `DataFrame`으로 행과 열 구조의 데이터를 표현


````python
import pandas as pd

# 딕셔너리 형태로 데이터 준비
data = {
    "이름": ["홍길동", "김철수", "이영희"],
    "나이": [25, 30, 28],
    "점수": [90, 85, 95]
}

# 딕셔너리를 DataFrame으로 변환
df = pd.DataFrame(data)

df
````
